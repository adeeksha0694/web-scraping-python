# PRODUCT PRICE WEB SCRAPER



```python
# Import libraries

from bs4 import BeautifulSoup
import requests
import smtplib
import time
import datetime
import csv
import pandas as pd
```

```python
# Connect to website

URL = 'https://www.web-scraping.dev/product/1'

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/153.0.0.0 Safari/537.36",
    "Accept-Encoding": "gzip, deflate",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "DNT": "1",
    "Connection": "close",
    "Upgrade-Insecure-Requests": "1"
}

CSV_FILE = 'ProductWebScrapperDataset.csv'
```

```python
# Fetch product information from website

page = requests.get(URL, headers=headers)

soup1 = BeautifulSoup(page.content, "html.parser")
soup2 = BeautifulSoup(soup1.prettify(), "html.parser")

title = soup2.find("h3", class_="product-title").get_text(strip=True)
price = soup2.find("span", class_="product-price").get_text(strip=True)

print(title)
print(price)
```

```python
# Clean product data

price = price.strip()[1:]
title = title.strip()

print(title)
print(price)
```

```python
# Create a timestamp for the output to track when data was collected

today = datetime.date.today()

print(today)
```

```python
# Create CSV file and store scraped data

header = ['Title', 'Price', 'Date']
data = [title, price, today]

with open(CSV_FILE, 'w', newline='', encoding='UTF8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerow(data)
```

```python
# Read CSV file using Pandas

df = pd.read_csv(CSV_FILE)

print(df)
```

```python
# Function to check the product price

def check_price():
    
    page = requests.get(URL, headers=headers)
    
    soup1 = BeautifulSoup(page.content, "html.parser")
    soup2 = BeautifulSoup(soup1.prettify(), "html.parser")
    
    title = soup2.find("h3", class_="product-title").get_text(strip=True)
    price = soup2.find("span", class_="product-price").get_text(strip=True)

    price = price.strip()[1:]
    title = title.strip()

    today = datetime.date.today()
    
    data = [title, price, today]

    with open(CSV_FILE, 'a+', newline='', encoding='UTF8') as f:
        writer = csv.writer(f)
        writer.writerow(data)

    print("Data added:")
    print(data)
```
